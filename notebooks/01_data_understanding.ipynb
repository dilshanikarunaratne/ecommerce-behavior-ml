Plotting the Correlation Map

```python
numeric_df = df.select_dtypes(include=['int64', 'float64'])

corr = numeric_df.corr()
```

```python
plt.figure(figsize=(10, 5))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

•	There is a very substantial positive correlation (0.86) between ProductRelated and ProductRelated_Duration. This implies that consumers who spend more time on product-related pages also prefer to visit such pages more frequently.

•	There is a very substantial positive connection (0.91) between BounceRates and ExitRates. In other words, individuals who quickly leave a page are also very likely to leave the website completely.

•	I chose to retain all of these columns for model development even though these variables may have overlapping data because of their high multicollinearity.

•	There is a moderately positive connection (0.60) between Administrative and Administrative_Duration. Consequently, consumers may spend more time on administrative sites if they visit them more frequently.

•	There is a positive connection (0.62) between Informational and Informational_Duration. People who visit more informational pages also read them for longer.

•	There are negative associations between ProductRelated and ProductRelated_Duration and: 

o	Bounce Rates 

o	Exit Rates 


•	It's possible that customers who actively peruse products are less likely to abandon the website right away.

•	The majority of other traits have weak connections with the following variables:

o	Operating Systems 

o	The browser 

o	Area 

o	Type of Traffic 


•	As a result, neither geographic nor technical factors significantly affect browsing behaviour on their o
wn.
•	The majority of variables have rather weak direct connections with PageValues. 

•	This could imply that a single behavioural parameter may not be the only factor influencing purchase intent.

•	It's crucial to remember that a weak association equals a useless feature. But for the initial iteration of my ML pipeline, I am maintaining all of these columns. Later, I will change the settings to see whether the results are improving.

In the next process of the ML pipeline, the model building and training should be first run in order to generate predictions in the next stage

The training model ("train.py") is located in "src" folder and the code is explained there.

Running the train.py file will create the models into "purchase_model.pkl" and segmentation of the online shoppers are done. However, to make the pipeline more reusable, the segmentation function is written in another file and the function is imported in the training model. ("segmentation.py")

High Intent Buyer: Users with very high purchase probability and strong engagement patterns.

Potential Buyer:Users showing moderate purchase intent and meaningful browsing behavior.

Engaged Browser: Users who actively browse products and spend significant time on the website but may not currently show strong purchasing probability.

At-Risk Visitor: Users with high bounce rates or exit rates, indicating weak engagement and potential drop-off behavior.

Low Intent Visitor: Users with low engagement and low purchase probability.
