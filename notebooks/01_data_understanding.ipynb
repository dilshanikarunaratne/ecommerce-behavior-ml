```python
import matplotlib.pyplot as plt
cols = [
    'Administrative', 'Administrative_Duration', 'Informational',
    'Informational_Duration', 'ProductRelated', 'ProductRelated_Duration',
    'BounceRates', 'ExitRates', 'PageValues', 'SpecialDay',
    'OperatingSystems', 'Browser', 'Region', 'TrafficType'
]

for col in cols:
    plt.figure(figsize=(4, 6))
    plt.boxplot(df[col])
    plt.title(col)
    plt.show()
```

```python
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1
print(IQR)
```

```python
import seaborn as sns

plt.figure()
sns.histplot(data=df, x='ProductRelated', hue='Revenue', bins=30, kde=True)
plt.title('Product Related Distribution')
plt.show()
```

```python
plt.figure()
sns.histplot(data=df, x='ProductRelated_Duration', hue='Revenue', bins=30, kde=True)
plt.title('Time Spent on Product Pages')
plt.show()
```

```python
plt.figure()
sns.histplot(data=df, x='PageValues', hue='Revenue', bins=30, kde=True)
plt.title('Distribution of Page Values')
plt.show()
```

```python
sns.scatterplot(
    data=df,
    x='ProductRelated_Duration',
    y='PageValues',
    hue='Revenue'
)
```

```python
sns.boxplot(x='Revenue', y='ProductRelated_Duration', data=df)
```

```python
sns.countplot(x='VisitorType', hue='Revenue', data=df)
```

```python
numeric_df = df.select_dtypes(include=['int64', 'float64'])

corr = numeric_df.corr()
```

```python
plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
corr_target = corr['Revenue'].sort_values(ascending=False)
print(corr_target)
```
